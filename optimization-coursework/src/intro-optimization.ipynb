## Practica 1: Primera toma de contacto con la optimización no lineal en python

#### 1.1 Existencia de solución

```python
import numpy as np
from scipy.optimize import minimize_scalar
```

```python
def f(x):
    return (np.tanh(x)- 1)**2
```

```python
res = minimize_scalar(f)
print(f"Solución x = {res.x}")
```

```python
res_2 = minimize_scalar(f, method='golden')
print(f"Solución x = {res_2.x}")
```

```python
import matplotlib.pyplot as plt

x = np.linspace(10, 50, 100)

plt.plot(x, f(x))
plt.show()
```

```python
x = np.linspace(17, 18, 100)

plt.plot(x, f(x))
plt.show()
```

```python
def f_reg(x):
    alpha = 1e-100  #hay que tunear
    return (np.tanh(x)-1)**2 + alpha*x**2
```

```python
res_reg = minimize_scalar(f_reg)
print(f"Solución x = {res_reg.x}")
```

```python
x = np.linspace(0, 20, 100)

plt.plot(x, f_reg(x))
plt.show()
```

#### Mínimos locales, máximos. inicialización, etc

```python
def objetivo(x):
    return (x**2 - 1)**2
```

```python
from scipy.optimize import minimize

x0 = 3 #Inicio de la iteración
res = minimize(objetivo, x0, method='BFGS',
               options={'gtol':1e-6, 'disp':True})

print(f"Solución x = {res.x[0]}")
```

```python
x = np.linspace(-1.5, 1.5, 100)

plt.plot(x, objetivo(x))
plt.show()
```
