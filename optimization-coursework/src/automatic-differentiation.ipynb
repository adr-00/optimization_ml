```python
import numpy as np
```

```python
# Implementamso la red neuranl vista en clase, y en la libreta
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivate(x):
    return sigmoid(x) * (1 - sigmoid(x))
```

```python
class NeuralNetwork:
    # Nota: Los pesos se inicializan manualmente solo con fines didácticos
    # para entender la diferenciación automática. En la práctica, se deben inicializar
    # aleatoriamente (por ejemplo, con np.random.randn) para permitir el aprendizaje.
    def __init__(self, input_size, hidden_size, output_size):
        self.weights_input_hidden = np.array([
            [1, 0],
            [-1, 1]
        ])

        self.weights_hidden_output = np.array([
            [1],
            [1]
        ])

    def mse_loss(self, y_true, y_pred):
        return np.mean((y_true - y_pred) ** 2)
    
    def forward(self, X):
        #First layer
        self.hidden_input = np.dot(X, self.weights_input_hidden) + self.bias_input_hidden
        print(f'self.hidden_input = {self.hidden_input}')
        self.hidden_output = sigmoid(self.hidden_input) 
        print(f'self.hidden_output = {self.hidden_output}')
        
        #Second layer
        self.final_input = np.dot(self.hidden_output, self.weights_hidden_output) + self.bias_hidden_output
        print(f'self.final_input = {self.final_input}')
        self.final_output = sigmoid(self.final_input)
        print(f'self.final_output = {self.final_output}')

        return self.final_output
    
    def compute_gradient(self, X, y_true):
        y_pred = self.forward(X)
        # error en la salida

        output_error = 2 * (y_pred - y_true) # derivada de c respecto de a2
        print(f'output_error = {output_error}')

        # gradiente de la capa de salida
        delta_output = output_error * sigmoid_derivate(self.final_input) # derivada de a2 respecto de z2
        print(f'delta_output = {delta_output}')
        grad_weights_hidden_output = np.dot(self.hidden_output.T, delta_output) # derivada de z2 respecto a w2

        # erro en la capa oculta
        hidden_error = np.dot(delta_output, self.weights_hidden_output.T) # derivada de z2 respecto a a
        print(hidden_error)
        delta_hidden = hidden_error * sigmoid_derivate(self.hidden_input) # derivada de a1 respecto a z1
        print(delta_hidden)
        grad_weights_input_hidden = np.dot(delta_hidden.T, X)
        print(np.shape(delta_hidden.T), np.shape(X))
        return grad_weights_hidden_output, grad_weights_input_hidden, y_pred
```

```python
X = np.array([[0],[0]])
y_true = np.array([[0]])

nn = NeuralNetwork(input_size=2, hidden_size=2, output_size=1)
grad_out, grad_in, y_pred = nn.compute_gradient(X=X.T, y_true=y_true)

print(f'funcion perdida = {nn.mse_loss(y_true, y_pred)}')
print(f'gradinte de los peses de la capa de salida = {grad_out}')
print(f'gradinte de los peses de la capa oculta = {grad_in}')
```
